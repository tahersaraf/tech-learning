## NumPy

NumPy is the foundational library for numerical computing in Python, providing efficient array operations and mathematical functions.

### Creating Arrays with `ndarray`

The central object in NumPy is the ndarray (n-dimensional array). You can create arrays from Python lists using `np.array()`, or generate them with helper functions like `np.zeros()`, `np.ones()`, and `np.arange()`

```python
import numpy as np

# Create an array from a Python list
a = np.array([1,2,3,4])
print("From list:", a)

# The size of the array with .size
print("Size of array:", a.size)

# Create a 2D array
b = np.array([[1,2,3],[4,5,6]])
print("2D array:\n",b)

# Dimension of an array with .ndim
print("Dimension of a:", a.ndim)
print("Dimension of b:", b.ndim)

# Useful creation functions
print("Zeroes:", np.zeros(4))
print("Ones:",np.ones(3))
print("Range:",np.arange(0,10))
```

### `dtype`

Every NumPy array has a `dtype` that describes the type of its elements (e.g., `int64`, `float64`, `bool`). You can inspect it with the `.dtype` attribute and specify it at creation time.

```python
# Defailt dtype is inferred from the data
int_array = np.array([1,2,3])
print("Integer array dtype:",int_array.dtype)

float_array = np.array([1.0,2.0,3.0])
print("Float array dtype:",float_array.dtype)

# Specify dtype explicitly
specified = np.array([1,2,3],dtype=np.float32)
print("Specified dtype:",specified.dtype,"| Values:", specified)
```

### Indexing, Slicing and Striding

NumPy supports the same indexing and slicing syntax as Python lists, extended to multiple dimensions. You can also use a **stride**(step size) to skip elements.

```python
arr = np.array([10,20,30,40,50,60,70,80])

# Basic Indexing
print("First Element:", arr[0])
print("Last Element:", arr[-1])

# Slicing [start:stop]
print("Elements 1-3:", arr[1:4])

# Striding [start:stop:step]
print("Every other element:", arr[::2])
print("Reversed:", arr[::-1])
```

### Aggregation

```python
grades = np.array([75, 80, 95, 90, 78, 82, 88, 91])

print("Sum of all elements: ", grades.sum())
print("Mean of all elements ", grades.sum())
print("Max value: ", grades.max())
print("Min value: ", grades.min())
print("Standard Deviation: ", grades.std())
```
