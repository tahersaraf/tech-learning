## Introduction to Classes

From W3Schools:
> *Python is an object oriented programming language. Almost everythin in Python is an object, with its properties and methods. A Class is like an object constructor, or a "blueprint" for creating objects.*

When working with functions, you can resuse the same logic by passing different arguments. Classes go a step further - they provide a reusable blueprint for both code and data.

To define a class, use the following syntax:

```py
class ClassName():
    <code block>
```
Unlike variables and functions, which use snake_case, class names follow the Python style guide(PEP 8) convention of CapWords - every word capitalized with no spaces

<div class="alert alert-success">
NOTE: The `pass` keyword tells Python to do nothing.
</div>

````python
class Dog():
    pass
````

### Instantiating a Class

For built-in types like `int` or `str`, Python creates objects automatically when you write a literal value such as `1` or `'hello'`. For custom classes you must explicitly create an object using the following syntax:

```py
object_name = ClassName()
```

This process is called **instantiation** - you are creating a specific insance of the class.

````python
# Instantiate the Dog class and assign it to the variable my_dog

my_dog = Dog()

type(my_dog)
````

### Functions are Objects

In Python, functions are first-class objects - just like integers, strings, and lists. This means you can assign a function a variable, pass it as an argument, or store it in a data structure.

When you assign a function to a variable, you use the function name without parentheses. Parentheses would *call* the function, without them, you are referencing the function object itself

```py
my_variable = some_function #assign the function (no parenthese)
my_variable()               #call it through the new name
```

````python
# Define a simple function
def greet(name):
    return f"Hello, {name}"

# Assign the function to a new variable
say_hello = greet

# Now both names point to the same function
print(say_hello("Rex"))
print(greet("Rex"))
print(type(say_hello))
````

### Defining Methods in a Class

Now that you can create a `Dog` object, lets add some fucntionality. Behavior in a class is defined using methods - special functions that belong to the class and are called on an object. 

```py 
object.method(args)
```
Defining a method is similiar to defining a regular function, with two key differences:
    1. The method definition is nested inside the class definition
    2. The first parameter must always be self, followed by additional parameters

The general structure looks like this:

```py
class ClassName():

    def method_name(self, args):
        # method code
```

````python
class UpdatedDog():

    def return_name(self,name):
        return f"name: {name}"
````

````python
my_updated_dog = UpdatedDog()

my_updated_dog.return_name("Rex")
````

### Understanding `self`

A method differs from a standalone function in that it can act on the specific object it was called on. The `self` parameter is how the method references that object.

When you call `object.method()`, Python automatically passes the object as the first argument to `self`. This means `self` always refers to the instance that invoked the method, regardless of what you named the variable

````python
class DogWithSelf():

    def print_self(self):
        print(self)

dog_with_self = DogWithSelf()

print(dog_with_self)

dog_with_self.print_self()
````

### The `__init__` method and Instance Attributes

In addition to methods, classes need a way to store data. **When a variable is stored on a class instance, it is called an attribute**. Each instance of a class has its own copy of the attributes, set to its own values.

Python provides a special method called `__init__(self)` that is automatically called when an object is created. This is known as the *constructor method* because it initializes the objects attributes at the moment of instantiation.

The general structure is:
```py
class ClassName():
    def __init__(self, arg):
        self.arg = arg
```

````python
class DogWithAttributes():
    def __init__(self, name, color):
        print("This ran automatically")
        self.name = name
        self.color = color

dog_with_attributes = DogWithAttributes("Rex","Brown")
````

When `__init__()` ran, it saved the values to `self`, binding them to the specific instance. You can access an attribute using dot notation - similiar to calling a method, but **without parenthese**

````python
dog_with_attributes.name, dog_with_attributes.color
````

Inside any method definition, you can access the object's attributes using `self.attribute_name`. Because `self` always refers to the calling instance, this works correctly regardless of what variable name was used during instantiation.

````python
class DogWithAttributesAndMethods():

    def __init__(self, name, color):
            self.name = name
            self.color = color

    def return_name(self):
          return self.name

    def return_color(self):
          return self.color

my_dog = DogWithAttributesAndMethods("Rex","Brown")

print(f"My dogs name is {my_dog.return_name()} and his coat is {my_dog.return_color()}")
````

Attributes are not fixed - they can be updated at any time. A method can modify an attribute by reassigning `self.attribute_name` to a new value.

````python
class DogWithAttributesAndMethods():

    def __init__(self, name, color):
            self.name = name
            self.color = color

    def return_name(self):
          return self.name

    def update_name(self, new_name):
          self.name = new_name

my_dog = DogWithAttributesAndMethods("Rex","Brown")
print(f"My dogs name is: {my_dog.return_name()}")

my_dog.update_name("Ruffles")
print(f"My dogs name after update is: {my_dog.return_name()}")
````

### Interacting with Other Class Instances

Classes can have many methods and attributes. They can also interact with other instances of the same class by accepting them as method arguments

````python
class DogFinal():

    def __init__(self, name_str, color_str):
        self.name = name_str
        self.color = color_str

    def return_name(self):
        return self.name

    def update_name(self, new_name):
        self.name = new_name

    def return_both_names(self, other_object):
        return self.name + " and " + other_object.name

dog_1 = DogFinal("Rex", "Brown")
dog_2 = DogFinal("Ruffles", "Black")

dog_1.return_both_names(dog_2)
````
