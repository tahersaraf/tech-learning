## Python Pandas

Importing library

```python
import pandas as pd
```

```python
data = [100,102,104]

series = pd.Series(data)

print(series)
```

```python
data = ['A','B','C']

series = pd.Series(data)

print(series)
```

### Series

```python
data = [100,102,104]

series = pd.Series(data, index=["a","b","c"])

print(series)
```

### loc - location

```python
print(series.loc["a"])
```

```python
series.loc["c"] = 200

series
```

### iloc - index location

```python
print(series.iloc[1])
```

### Conditionals

```python
print(series[series >= 200])
```

```python
print(series[series < 200])
```
