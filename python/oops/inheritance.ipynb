## Inheritance

**Inheritance** allows you to create a new class (called a **child** or **subclass**) that automatically receives all the attributes and methods of an existing class (called a **parent** or **superclass**) . The child can then add new behavior or override existing behavior without modifying the parent.

This is useful when you have a general concept (e.g. Animal) and want to create more specific versions of it (e.g., Dog, Cat) that share common logic but differ in certain details

The syntax for inheritance is:
```py
class ChildClass(ParentClass):
    <code block>
```

You pass the parent class name inside the parenthese of the child class definition. The child inherits everything from the parent and can:
* Use parent methods and attributes
* Override parent methods by redifining them
* Extend the parent with new methods and attributes

User `super().__init__()` inside the childs `__init__` to call the parents constructor and initialize inherited attributes

````python
# Parent Class
class Animal():
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says {self.sound}"

# Child Class
class Dog(Animal):
    def __init__(self, name, color):
        super().__init__(name, sound="Woof") # Call parents __init__
        self.color = color                   # Add new attribute

    def fetch(self):
        return f"{self.name} fetches the ball"

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, sound="Meow")

my_dog = Dog("Rex","Brown")
my_cat = Cat("Garfield")

print(my_dog.speak())
print(my_cat.speak())
print(my_dog.name)
print(my_dog.color)
print(my_cat.name)
    
````

### Multiple Inheritance

Multiple inheritance in Python allows a class to inherit attributes and methods from more than one parent class, enabling complex relationships and code reuse.

In Python, a class can inherit from multiple parent classes by listing them in parentheses:

````python
class Mammal:
    def mammal_info(self):
        print("Mammals can give direct birth.")
        
class WingedAnimal:
    def winged_animal_info(self):
        print("Winged animals can flap.")

class Bat(Mammal, WingedAnimal):
    pass

b1 = Bat()
b1.mammal_info()          # Output: Mammals can give direct birth.
b1.winged_animal_info()   # Output: Winged animals can flap.
````

### Diamond Problem and Method Resolution Order (MRO)

The diamond problem occurs when two parent classes inherit from the same base class, and a child class inherits from both, creating ambiguity in method calls. Python resolves this using Method Resolution Order (MRO):

````python
class A:
    def show(self):
        print("Class A")
        
class B(A):
    def show(self):
        print("Class B")

class C(A):
    def show(self):
        print("Class C")

class D(B, C):
    pass


d = D()
d.show()  # Output: Class B
print(D.mro())  # Output: [D, B, C, A, object]
````
