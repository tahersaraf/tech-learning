```python
import pandas as pd
```

### Creating a DataFrame

`pandas` is a library that provides data types and functions for rigorous, programmatic data analysis. The core `pandas` data type is the DataFrame.

A `DataFrame` is a two-dimensional table of named rows and columns, similiar to a SQL table. The DataFrame class accepts a data attribute that defines the content of the table at instantiation

To create a DataFrame, you can pass a list of lists where each inner list represents a row of data

|Name | Age | Job
|---|---|---|
|John | 30 | Journalist |
|Mary | 30 | Programmer |
|Abe | 40 | Chef|

```python
data = [["John",30,"Journalist"],["Mary",30,"Programmer"],["Abe",40,"Chef"]]

df = pd.DataFrame(data=data)
df
```

### Adding Column Names

The column headers `0, 1, 2` shown above are default integer values assigned when no columns names are provided. To assign meaningful column names, pass a list of strings to the columns attribute of `pd.Dataframe()`.

```python
cols = ["Name","Age","Job"]

df = pd.DataFrame(data=data, columns=cols)
df
```

```python
df.dtypes
```

### Creating DataFrame from a NumPy ndarray

You can also create a Dataframe from a NumPy array. This is useful when your data is already in array form, such as after numerical computations or loading from external sources. Simply pass the NumPy array as the data argument and specify column names with the columns parameter.

```python
import numpy as np

np_array = np.array([
    ["John",30,"Journalist"],
    ["Mary",30,"Programmer"],
    ["Abe",40,"Chef"]
    ])

df2 = pd.DataFrame(data=np_array, columns=cols)
df2
```

```python
df2.dtypes
```

<div class="alert alert-info">
Note:
When creating a DataFrame from a NumPy array containing mixed types, NumPy assigns the 'object' dtype to all elements, so pandas infers all columns as object type
</div>

### The Series Datatype

The other primary data type in `pandas` is the Series. A Series represents a single column of a Dataframe.

There are two ways to select a Series from a Dataframe:
1. `df["column_name"]` -- bracket indexing (preferred)
2. `df.column_name` -- attribute reference

It is preferred to use `df["column_name"]` because the attribute reference notation does not work when column names contain spaces

```python
df["Age"]
```

```python
df.Age
```

```python
df2.Age
```

### `dtypes` and `Series` Operations

When you display a `Series`, you will notice a `dtype` label at the bottom (e.g., `dtype: int64`). In `pandas` `dtypes` describes the type of values stored in a column.

Just as Python's built in types determine what operations are valid, the dtype of a `pandas` column determines what operations can be performed on it. For example, you can compute the average of a numeric column, but not a text column

| `pandas` dtype | Description | Python equivalent |
|---|---|---|
| `object` | Text/string data | `str` |
| `int64` | Integer values | `int` |
| `float64` | Floating-point values | `float` |
| `bool` | Boolean values | `bool` |

Operations on a Series behave similarly to operations on their Python counterpart types, but they act **element-wise** -- meaning the operation is applied to every value in the column

For example, adding two int64 Series together produces a new Series where each element is the sum of the corresponding elements

```python
df["Age"] + df["Age"]
```

```python
df["Age"] + 3 - 1
```

### Selecting a Single Value

To retrieve a single value from a Series, use integer-based notation - the same syntax used to index into a Python list. The example below retrieves the first value from the Age colum.

```python
df["Age"][0]

print(df["Age"][0])
```

### Selecting Multiple Columns

In addition to selecting a single column as a Series, you can select a subset of columns and return the result as a Dataframe.

```python
df[["Name","Age","Job"]]
```

### Using `iloc` for integer-based indexing

The `iloc` method allows you to select rows and columns by their integer positions.

The syntax is:
`df.iloc[row_selection, column_selection]`

In the expression df.iloc[:, [0,2]], the colon(:) before the comma means "select all rows" while [0,2] after the comma specifies the first and third columns. This is useful when you want to select columns by their order rather than by name


```python
# Get the original dataframe
df
```

```python
# Get the first and third column only
df.iloc[:,[0,2]]
```

```python
# Get the first row as a dataframe

df.iloc[:1,:]
```

```python
# Get the first row as a series

df.iloc[0, :]
```

<div class="alert alert-info">
<b>Note:</b>

Examples of how to use iloc to select specific rows and columns by their integer positions:

* `iloc[0, 0]` gets the first cell -- 'John'
* `iloc[:, 0]` gets the entire first column
* `iloc[0, :]` gets the entire first row

</div>

### Mutability of DataFrame Objects

When you perform an operation that modifies a DataFrame - such as dropping  a column - pandas returns a new object by default, leaving the original DataFrame unchanged. This behavior makes it safer to build complex data pipelines, since intermediate transformers do not unintentionally overwrite source data. 

The cell below drops the `Age` column and displays the result. Because the result is not assigned to a variable, it is discarded after display.

```python
df.drop("Age", axis=1)
```

The original DataFrame df is unchanged, as confirmed by displaying it again. To persist the result of an operation like drop(), assign the returned object to a new variable.

```python
df
```

```python
df_no_age = df.drop("Age", axis=1)

df_no_age
```

<div class="alert alert-info">
<b>Note:</b>

The 'axis' parameter specifies whether the operation applies to rows or columns:
* axis = 1 means to operate on columns (e.g., drop a column)
* axis = 0 means to operate on rows (e.g., drop a row)
</div>

### In-Place Modification

When you want to modify a DataFrame directly - without creating a new object - you can pass `inplace=True` to the operation. In this case, no new object is returned; the change is applied to the original DataFrame

```python
df.drop("Age", axis=1, inplace=True)
```

```python
# The Age columns has been dropped from the original DataFrame

df
```

### Working with Date Columns

When `pandas` reads a CSV file, date columns are typically loaded as plain strings. To perform any date-based operations - such as sorting by date, calculating time differences, or extracting year/month values - the column must first be converted to a proper datetime64 type.

The `pd.to_datetime()` function parses a Series of date strings into datetime objects. We use the format parameter to tell exactly how the input strings are structured.

```python
df["start_date"]
```

```python
df["start_date"] = pd.to_datetime(df["start_date"])
```

```python
df[["name","start_date"]].head(10)
```

```python
df[["name","start_date"]].dtypes
```

### Reformate Dates with `strftime()`

Once a column has been converted to `datetime64`, we can reformat the date values into any string respresentation we need using `strftime()` (shor for string format time).

If we want to display dates in MM-DD-YYYY format (e.g., 07-23-2009). We achieve this with the format string `"%m-%d-%Y"`

<div class="alert alert-info">
    <b>Note:</b>

`strftime()` converts the column back to object (string) dtype. If you need to keep the column as `datetim64` for further date operations, store the formatted version in a seperate column.
</div>

```python
df["start_date"] = df["start_date"].dt.strftime("%m-%d-%Y")
```

```python
df.head(3)
```

### Renaming Columns

We can rename columns of a DataFrame using `rename()`. We pass a dictionary to the columns parameter that maps old column names to new ones.

The syntax for renaming one column is:
df = df.rename(columns={"old_name": "new_name"})

The syntax for renaming more than one column is:
df.rename(columns={"old_col_a": "new_col_a", "old_col_b": "new_col_b:})

```python
df = df.rename(columns={"name": "Name"})
```

```python
df.head(3)
```

### Filtering

To select a subset of rows that meet a certain condition, use the syntax `df[bool_array]`, where `bool_array` is a series of True and False values - one per row

* Rows that evaluate to True are kept
* Rows that evaluate to False are dropped
