## Errors and Exceptions

### Syntax Errors

In Pythion, there are two main categories of errors: **Syntax Errors** and **Exceptions**. A Syntax Error occurs when code is written in a way that Python cannot interpret - the code violates the rules of the language itself so Python cannot even attempt to run it.

````python
if print("Hello Wordl")  # This is not correct Python code, so it throws a Syntax Error
````

### Exceptions

When Python can parse and understand the code but encounters a problem while executing it, the resulting error is called an **Exception**. Unlike Syntax Errors, exceptions occur at runtime - Python understood what you were trying to do, but something went wrong during execution. 

The example below triggers a ZeroDivisionError, one of Pythons built-in exception types.

````python
1 / 0
````

### Built-In Exception Types

Python provides a rich set of built-in exceptions, each indicating a specific category of problem. The ZeroDivisionError seen above signals an attempt to divide by zero.

Here are some of the most common exceptions.

* SyntaxError - Raised when code violates Python's grammer rules (e.g., missing colo, unmatched parentheses)
* NameError - Raised when referencing a variable or function that hasn't been defined
* TypeError - Raised when an operation is applied to an object of inappropriate type (e.g., "hello" + 5)
* ValueError - Raised when an operation receives an arguement of the right type but inappropriate value (e.g., int("abc"))
* IndexError - Raised when accessing list or tuple index that doesn't exist
* KeyError - Raised when accessing a dictionary key that doesn't exist
* ZeroDivisionError - Raised when dividing by zero

## Exception Handling

### The `try / except` Statement

Syntax Errors will always cause a program to exit, but **Exceptions can be handled**. Python's `try/except` statement lets you define what should happen when an exception is raised, rather than allowing the program to crash.

The basic structure is:

```py
try:
    # code that may raise an exception
except:
    # code to run if exception occurs
```
When Python enters a try block and encounters an exception, it immidiately jumps to the except block instead of exiting with an error. The code below demonstrates this:

````python
try:
    1/0
except:
    print("Exception Handled")
````

### Handling a Specific Exception

The broad except clause catches any exception. For more precise control, you can specify the exception type you want handle by naming it after the except keyword:

```py
try:
    # code that may raise an exception
except ExceptionName:
    # code to run only if ExceptionName is raised
```

In the first example below, a ZeroDivisionError is raised and matched by the handler. In the second example a NameError is raised instead - because the handler only targets ZeroDivisionError, the exception would go unhandled.

````python
try:
    1/0
except ZeroDivisionError:
    print("Exception Handled")
````

````python
try:
    print(undefined_variable)
except ZeroDivisionError:
    print("Exception Handled")
````

### Hanlding Multiple Exceptions

To handle more than one specific exception type in a single except clause, list the exception names in a tuple inside parenthese:

```py
except (ExceptionType1, ExceptionType2):
    # handles either exception
```

````python
try:
    1/0
    print(undefined_variable)
except (ZeroDivisionError,NameError):
    print("Exception Handled")
````

### Printing the Exception Error

Printing the exception error helps you understand exactly what went wrong, making debugging and troubleshooting easier.

````python
try:
    # 1/0
    print(undefined_variable)
except (ZeroDivisionError,NameError) as e:
    print(f"Exception Handled: {e}")
````

### The `else` Clause

You can add an else block after except to define code that runs **only when no exception was raised** inside the try block:

```py
try:
    # code that may raise an exception
except ExceptionType:
    # runs if exception occurs
else:
    # runs only if NO exception occurs
```

This is useful for seperating the "happy path" logic from error handling

````python
try:
    result = 10/2
except ZeroDivisionError:
    print("Cannot Divide by Zero")
else:
    print(f"Divison successful! Result: {result}")
````

### The `finally` Clause

The finally block runs **regardless of whether an exception occured or not**. It is typically used for cleanup actions - such as closing files or releasing resources - that must happen no matter what:

```py
try:
    # code that may raise an exception
except ExceptionType:
    # runs if exception occurs
finally:
    # ALWAYS runs, exception or not
```

````python
try:
    result = 10/0
except ZeroDivisionError:
    print("Cannot Divide by Zero")
else:
    print(f"Divison successful! Result: {result}")
finally:
    print("This always executes - use it for cleanup tasks.")
````

### Assert Statements

The `assert` statement is a concise way to raise an AssertionError when a condition is not met. It is commonly used for validation and debugging:

```py
assert boolean_expression, optional message
```
When Python evaluates ans assert statement
* If the boolean expression is **True**, Python does nothing and moves on
* If the boolean expression is **False**, Python raises an AssertionError with the optional mesage if one was provided

````python
assert 1==1
````

````python
assert 1==2, "That is not true"
````
