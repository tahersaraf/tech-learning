## Python Pandas

Importing library

```python
import pandas as pd
```

```python
data = [100,102,104]

series = pd.Series(data)

print(series)
```

```python
data = ['A','B','C']

series = pd.Series(data)

print(series)
```

### Series

```python
data = [100,102,104]

series = pd.Series(data, index=["a","b","c"])

print(series)
```

### loc - location

```python
print(series.loc["a"])
```

```python
series.loc["c"] = 200

series
```

### iloc - index location

```python
print(series.iloc[1])
```

### Conditionals

```python
print(series[series >= 200])
```

```python
print(series[series < 200])
```

```python
calories = {"Day 1": 1750, "Day 2": 2100, "Day 3": 1700}

series = pd.Series(calories)

print(series)
```

```python
series.loc["Day 3"] += 500

series.loc["Day 3"]
```

### DataFrame
Tabular form of data

```python
data = {"Name": ["Spongebob","Patrick","Squidward"],
        "Age": [30,35,50]}

df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"])

print(df)
```

```python
df.loc["Employee 1"]
```

```python
df.iloc[1]
```

### Add a new column

```python
df["Job"] = ["Cook","N/A","Cashier"]

df
```

```python
new_row = pd.DataFrame([{"Name": "Sandy", "Age": 28, "Job": "Engineer"}],['Employee 4'])

pd.concat([df, new_row])
```

#### Filtering

```python
print(df.to_string())
```

```python
df = pd.read_csv("sample-employees.csv")

highest_earner = df[df["salary"] > 150000]

print(highest_earner.to_string())
```

```python
desgin_team = df[df["department"]=="Design"]

print(desgin_team.to_string())
```

#### Agregate Function - `groupby()`

```python
df = pd.read_csv("sample-employees.csv")

print(df.to_string())
```

```python
print(df.mean(numeric_only=True))
```

```python
print(df.sum(numeric_only=True))
```

```python
print(df.min(numeric_only=True))
```

```python
print(df.max(numeric_only=True))
```

```python
print(df.count())
```

```python
group = df.groupby("department")

print(group["salary"].max())
```
