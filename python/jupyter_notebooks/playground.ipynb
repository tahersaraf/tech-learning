```python
count = {}
s = "hello world"

for char in s:
    count[char] = count.get(char, 0) + 1

for char, num in count.items():
    if num > 1:
        print(char, num)
```

```python
class Solution:
    def reverseString(self, s: list[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        l, r = 0, len(s) - 1

        while l < r:
            s[l], s[r] = s[r], s[l]
            l += 1
            r -= 1
        return s

sol = Solution()

print(sol.reverseString(["h","e","l","l","o"]))
```

```python

def isValid(s: str) -> bool:
    stack = []
    pairs = {"}":"{","]":"[",")":"("}

    for i in s:
        if i in pairs:
            if not stack or stack[-1] != pairs[i]:
                return False
            stack.pop()
        else:
            stack.append(i)
    return len(stack) == 0

print(isValid("()"))
print(isValid("()[]{}"))
print(isValid("(]"))
print(isValid("([])"))
```
